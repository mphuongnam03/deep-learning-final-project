# TBX11K Dataset and Training Analysis

Notebook này đọc dataset `tbx11k-simplified/data.csv`, phân tích phân bố dữ liệu, đọc training metrics và sinh biểu đồ vào `reports/plots/`.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.analytics.reporting import add_class_name, generate_report_plots
import pandas as pd

dataset_csv = ROOT / 'tbx11k-simplified' / 'data.csv'
df = add_class_name(pd.read_csv(dataset_csv))
df.head()
```

```python
summary = {
    'rows': len(df),
    'columns': list(df.columns),
    'target_distribution': df['target'].value_counts().to_dict(),
    'image_type_distribution': df['image_type'].value_counts().to_dict(),
    'class_distribution': df['class_name'].value_counts().to_dict(),
    'bbox_distribution': df['bbox'].apply(lambda x: 'has_bbox' if pd.notna(x) and x != 'none' else 'no_bbox').value_counts().to_dict(),
}
summary
```

```python
created = generate_report_plots(
    dataset_csv=str(ROOT / 'tbx11k-simplified' / 'data.csv'),
    detection_results_csv=str(ROOT / 'tb_detection' / 'stage2_det' / 'results.csv'),
    classification_results_csv=str(ROOT / 'tb_classification' / 'stage1_cls' / 'results.csv'),
    output_dir=str(ROOT / 'reports' / 'plots'),
)
created
```
